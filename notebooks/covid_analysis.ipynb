```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df_day = pd.read_csv('day_wise.csv')
df_country = pd.read_csv('country_wise_latest.csv')
print(df_day.head())
print(df_country.head())
```

```python
plt.figure(figsize=(12,5))
plt.plot(df_day['Date'], df_day['Confirmed'], color='red')
plt.title('Daily COVID-19 Cases Trend')
plt.xlabel('Date')
plt.ylabel('Confirmed Cases')
plt.xticks(df_day['Date'][::30], rotation=45)
plt.tight_layout()
plt.savefig('daily_trend.png')
plt.show()
```

```python
top10 = df_country.nlargest(10, 'Confirmed')[['Country/Region','Confirmed','Deaths','Recovered']]
top10.plot(kind='bar', x='Country/Region', figsize=(12,5))
plt.title('Top 10 Countries - Region Wise Comparison')
plt.tight_layout()
plt.savefig('region_comparison.png')
plt.show()
```

```python
df_country['Death Rate'] = (df_country['Deaths'] / df_country['Confirmed']) * 100
df_country['Recovery Rate'] = (df_country['Recovered'] / df_country['Confirmed']) * 100
avg = df_country[['Death Rate', 'Recovery Rate']].mean()
avg.plot(kind='bar', color=['red','green'], figsize=(6,4))
plt.title('Average Death vs Recovery Rate')
plt.tight_layout()
plt.savefig('death_recovery.png')
plt.show()
```
